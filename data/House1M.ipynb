```python
import pandas as pd
import numpy as np

# df = pd.read_csv("House1.csv",
#     index_col='Date_Time',
#     parse_dates=True
# )

df=pd.read_csv("House1.csv")
print(df.head())
df.info()
print(df.isnull().sum)
```

```python
columns_to_delete = [
    'AC_DR_kW', 
    'UPS_kW', 
    'LR_kW', 
    'Kitchen_kW', 
    'AC_Dr_kW', 
    'AC_BR_kW'
]


df.drop(columns=columns_to_delete, inplace=True)


print(df.head())
```

```python

df['Date_Time'] = pd.to_datetime(df['Date_Time'])


df = df.set_index('Date_Time')


df_daily = df['Usage_kW'].resample('D').sum() * (1/60) 

df_daily = df_daily.to_frame()
df_daily = df_daily.rename_axis('Date')
df_daily.columns = ['Usage_kWh']


print(df_daily)
```

```python

excel_file_name = 'House1.csv'
df_daily.to_csv(excel_file_name)

print(f"File '{excel_file_name}' Excel format mein save ho gaya hai.")
```
